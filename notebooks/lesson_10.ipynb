# Python：`input()` 與 `split()` 用法

這份筆記介紹 Python 中最常見的輸入方式：

1. `input()`：讀取一整行文字
2. `split()`：把一行文字切成多個部分
3. `map(int, ...)`：把輸入的文字轉成整數
4. 搭配 `for`：重複讀取多行資料

> 重點：`input()` 讀進來的資料預設都是 **字串（string）**。


## 1. `input()`：讀取一整行

最基本的寫法：

```python
name = input()
```

執行後，程式會等待使用者輸入一行資料。

例如輸入：

```text
Jasper
```

變數 `name` 就會得到字串 `"Jasper"`。


````python
name = input("請輸入名字：")
print("你輸入的是：", name)
````

也可以直接在 `input()` 裡加入提示文字：

```python
age = input("請輸入年齡：")
```

但要注意：即使輸入 `18`，得到的仍然是字串 `"18"`。


````python
age = input("請輸入年齡：")

print(age)
print(type(age))
````

## 2. 把輸入轉成整數

如果要做數學運算，可以用 `int()`：

```python
age = int(input())
```

例如輸入：

```text
18
```

此時 `age` 就是整數 `18`。


````python
age = int(input("請輸入年齡："))

print(age + 1)
````

## 3. `split()`：把一行切成多個資料

假設輸入：

```text
10 20 30
```

如果只使用：

```python
input()
```

會得到一整個字串：

```text
"10 20 30"
```

如果使用：

```python
input().split()
```

會依照空白切開，得到：

```python
["10", "20", "30"]
```


````python
data = input("請輸入三個資料（用空白隔開）：").split()

print(data)
print(type(data))
````

### `split()` 預設以空白分隔

```python
words = input().split()
```

例如輸入：

```text
apple banana orange
```

結果：

```python
["apple", "banana", "orange"]
```

因此可以利用索引取得不同位置的資料。


````python
words = input("請輸入三個單字：").split()

print("第一個：", words[0])
print("第二個：", words[1])
print("第三個：", words[2])
````

## 4. 一次存入多個變數

如果確定輸入剛好有兩個資料，可以寫：

```python
x, y = input().split()
```

例如輸入：

```text
hello world
```

則：

```python
x == "hello"
y == "world"
```


````python
x, y = input("請輸入兩個字串：").split()

print("x =", x)
print("y =", y)
````

## 5. 一次讀入多個整數

競賽題目很常出現：

```text
10 20
```

我們希望 `x` 和 `y` 都直接變成整數。

可以使用：

```python
x, y = map(int, input().split())
```

執行順序可以想成：

```text
input()
   ↓
"10 20"
   ↓ split()
["10", "20"]
   ↓ map(int, ...)
10, 20
```


````python
x, y = map(int, input("請輸入兩個整數：").split())

print("x + y =", x + y)
````

### 三個以上的整數也一樣

```python
a, b, c = map(int, input().split())
```

例如輸入：

```text
3 5 7
```


````python
a, b, c = map(int, input("請輸入三個整數：").split())

print("總和 =", a + b + c)
````

## 6. 不知道有幾個整數時：轉成 `list`

如果一行中整數的數量不固定，可以寫：

```python
nums = list(map(int, input().split()))
```

例如輸入：

```text
1 2 3 4 5
```

會得到：

```python
[1, 2, 3, 4, 5]
```


````python
nums = list(map(int, input("請輸入數個整數：").split()))

print(nums)
print("總和 =", sum(nums))
````

## 7. 自訂分隔符號

`split()` 不一定只能用空白。

例如：

```python
text = "apple,banana,orange"
text.split(",")
```

會得到：

```python
["apple", "banana", "orange"]
```


````python
text = "apple,banana,orange"

fruits = text.split(",")

print(fruits)
````

## 8. 使用 `for` 重複讀取多行

原本的寫法：

```python
for i in range(4):
    input()
```

代表程式會讀取 **4 行輸入**，但沒有把資料存下來。

如果要保留每一行，可以加入變數。


````python
for i in range(4):
    text = input(f"請輸入第 {i + 1} 行：")
    print("讀到：", text)
````

如果每一行都有兩個整數，例如：

```text
1 2
3 4
5 6
7 8
```

可以寫：


````python
for i in range(4):
    x, y = map(int, input().split())
    print("x =", x, "y =", y)
````

## 9. 常見錯誤

### 錯誤 1：忘記轉成整數

```python
x, y = input().split()
print(x + y)
```

如果輸入：

```text
10 20
```

輸出會是：

```text
1020
```

因為 `x` 和 `y` 都是字串。

正確寫法：

```python
x, y = map(int, input().split())
```


### 錯誤 2：變數數量和輸入數量不同

```python
x, y = input().split()
```

如果輸入三個值：

```text
10 20 30
```

就會發生錯誤，因為左邊只有兩個變數。

如果資料數量不固定，可以改用：

```python
data = input().split()
```


### 錯誤 3：把 `split()` 寫在數字上

不可以：

```python
int(input()).split()
```

因為 `split()` 是字串的方法。

通常應寫成：

```python
map(int, input().split())
```

也就是先切開字串，再把每一個部分轉成整數。


## 10. 常用寫法整理

| 需求 | 寫法 |
|---|---|
| 讀取一行文字 | `s = input()` |
| 讀取一個整數 | `n = int(input())` |
| 讀取多個字串 | `data = input().split()` |
| 讀取兩個字串 | `a, b = input().split()` |
| 讀取兩個整數 | `a, b = map(int, input().split())` |
| 讀取數量不固定的整數 | `nums = list(map(int, input().split()))` |
| 重複讀取 `n` 行 | `for i in range(n): ...` |

## 小結

看到輸入題目時，可以先判斷：

- **一整行文字** → `input()`
- **一行多個資料** → `input().split()`
- **一行多個整數** → `map(int, input().split())`
- **數量不固定的整數** → `list(map(int, input().split()))`
- **有很多行資料** → 搭配 `for` 迴圈

最常用的一行：

```python
a, b = map(int, input().split())
```
