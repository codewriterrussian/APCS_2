```python
n = int(input())

counts = [0,0,0]

for i in range(n):
    num = int(input())
    remainder = num % 3
    counts[remainder] += 1

print(counts[0], counts[1], counts[2])
```

```python
rows, cols = map(int, input().split())

matrix = []

for i in range(rows):
    row = list(map(int, input().split()))
    matrix.append(row)

for j in range(cols):
    result = []

    for _ in range(rows):
        result.append(matrix[i][j])
    
    print(*result)
```
