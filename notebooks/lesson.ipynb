```python
tokens = input().split()
stack = []

# f,f,f,2

for token in reversed(tokens):
    if token == "f":
        x = stack.pop()
        stack.append(2 * x - 3)
    else:
        stack.append(int(token))
```

```python
def power(role):
    attack = role[0]
    defense = role[1]
    return attack ** 2 + defense ** 2

n = int(input())
roles = []
for _ in range(n):
    role = list(map(int, input().split()))
    roles.append(role)


# (1, power(1)), (2, power(2)), (3, power(3))
# sort by power, descending -> (2, power(2)), (1, power(1)), (3, power(3))
roles.sort(key=power, reverse=True)
print(roles[1][0], roles[1][1])
```

```python
def f(x):
    return 2 * x - 3
def g(x, y):
    return 2 * x + y - 7
def h(x, y, z):
    return 3 * x - 2 * y + z

tokens = input().split()

def calculate():
    token = tokens.pop(0)
    if token == "f":
        return f(calculate())
    if token == "g":
        return g(calculate(), calculate())
    if token == "h":
        return h(calculate(), calculate(), calculate())
    return int(token)

print(calculate())
```
